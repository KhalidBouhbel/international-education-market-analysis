```python
df_EdStatsData_1['NB_NAN'] = df_EdStatsData_1.isna().sum(axis=1)

df_EdStatsData_1 = df_EdStatsData_1.sort_values('NB_NAN')

df_EdStatsData_1 = df_EdStatsData_1.drop_duplicates('Country Name', keep='first')

df_EdStatsData_1 = df_EdStatsData_1.drop('NB_NAN', axis=1)
```

```python
df_EdStatsData_1['SCORE_1'] = df_EdStatsData_1['2011'] / df_EdStatsData_1['2011'].max()
```

```python
df_EdStatsData_1['SCORE_2'] = df_EdStatsData_1['2013'] / df_EdStatsData_1['2013'].max()
```

```python
df_EdStatsData_1['SCORE_3'] = df_EdStatsData_1['2016'] / df_EdStatsData_1['2016'].max()
```

```python
df_EdStatsData_1[['SCORE_1', 'SCORE_2']].corr()
```

```python
df_EdStatsData_1[['SCORE_1', 'SCORE_3']].corr()
```

```python
df_EdStatsData_1[['SCORE_2', 'SCORE_3']].corr()
```

```python
df_EdStatsData_1['SCORE_SYNT'] = (df_EdStatsData_1['SCORE_1'] + df_EdStatsData_1['SCORE_2'] +  df_EdStatsData_1['SCORE_3']) / 3
```

```python
df_EdStatsData_definitif = df_EdStatsData_1.sort_values('SCORE_SYNT', ascending=False)[['Country Name', 'Country Code', 'Indicator Name','Indicator Code', 'SCORE_1', 'SCORE_2', 'SCORE_3', 'SCORE_SYNT']].iloc[:15]
```
