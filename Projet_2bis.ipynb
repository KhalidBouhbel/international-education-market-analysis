```python
# utilisation de fonctions pour réutiliser plusieurs fois un même programme 
```

```python
def plot_tendancy(df_indic, title, ylabel):
    
    dfR1 = df_indic.loc[df_indic['Country Name']=='Norway', :] 
    dfR2 = df_indic.loc[df_indic['Country Name']=='Luxembourg', :]
    dfR3 = df_indic.loc[df_indic['Country Name']=='Sweden', :]
    dfR4 = df_indic.loc[df_indic['Country Name']=='Iceland', :]
    dfR5 = df_indic.loc[df_indic['Country Name']=='Netherlands', :]
    
    x = ["2012", "2013", "2014", "2015", "2016", "2017"]
    y1 = [dfR1['2012'], dfR1['2013'], dfR1['2014'], dfR1['2015'], dfR1['2016'], dfR1['2017']]
    y2 = [dfR2['2012'], dfR2['2013'], dfR2['2014'], dfR2['2015'], dfR2['2016'], dfR2['2017']]
    y3 = [dfR3['2012'], dfR3['2013'], dfR3['2014'], dfR3['2015'], dfR3['2016'], dfR3['2017']]
    y4 = [dfR4['2012'], dfR4['2013'], dfR4['2014'], dfR4['2015'], dfR4['2016'], dfR4['2017']]
    y5 = [dfR5['2012'], dfR5['2013'], dfR5['2014'], dfR5['2015'], dfR5['2016'], dfR5['2017']]
    plt.plot(x,y1 , label = 'Norvège')
    plt.plot(x,y2, label = 'Luxembourg')
    plt.plot(x,y3, label = 'Sweden')
    plt.plot(x,y4, label='Iceland')
    plt.plot(x,y5, label='Netherlands')
    plt.title(title, fontsize = 18 , fontweight='bold' )
    plt.xlabel("Année" , color = 'blue' , fontsize = 16)
    plt.ylabel(ylabel , color = 'blue' , fontsize = 16)
    plt.legend()
    plt.gcf().set_size_inches(12, 6)
```

```python
plot_tendancy(df_RNB1, "Evolution RNB", "RNB ($/hab)")
```

```python
plot_tendancy(df_NiveauEduc1, "Evolution niveau éduc", "Taux")
```
